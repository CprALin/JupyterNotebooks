```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules 
```

```python
data = pd.read_csv('./l34.csv')
data.head()
```

```python
data.columns
```

```python
data.Item.unique()
```

```python
data.info()
```

```python
data_encoded = data.pivot_table(index="Basket" , columns="Item" , aggfunc=lambda x:1 , fill_value = 0)
data_encoded = data_encoded.astype(bool)
```

```python
freq_items = apriori(data_encoded , min_support=0.05 , use_colnames=True)
rules = association_rules(freq_items , metric="lift" , min_threshold= 1)
rules = rules.sort_values(['confidence','lift'] , ascending=[False,False])
print(freq_items)
rules.head()
```
