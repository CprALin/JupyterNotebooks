### Steps in modelling 

The steps in modelling with TensorFlow are typically :

1. Create or import a model 
2. Compile the model
3. Fit the model 
4. Evaluate the model


```python
import tensorflow as tf
# Set the random seed 
tf.random.set_seed(42)

# 1. Create the model using the Sequential API 
model_1 = tf.keras.Sequential([
    tf.keras.layers.Dense(1)
])

# 2. Compile the model
model_1.compile(loss=tf.keras.losses.BinaryCrossentropy(),
                optimizer=tf.keras.optimizers.Adam(),
                metrics=["accuracy"])

# 3. Fit the model
model_1.fit(X, y , epochs=100)
```

```python
# Let's try and imrpove our model by training for longer ...
model_1.fit(X, y , epochs=250)
model_1.evaluate(X, y)
```

Model is getting ~50% accuracy (this is bad)

Let's improve it

```python
# Set random seed
tf.random.set_seed(42)

# 1. Create the model (with 2 hidden layers)
model_2 = tf.keras.Sequential([
    tf.keras.layers.Dense(1),
    tf.keras.layers.Dense(1)
])

# 2. Compile the model
model_2.compile(loss=tf.keras.losses.BinaryCrossentropy(),
                optimizer=tf.keras.optimizers.Adam(),
                metrics=["accuracy"])

# 3. Fit the model
model_2.fit(X, y , epochs=150)
```

```python
model_2.evaluate(X, y)
```

Create a function `plot_decision_boundary()` , this funtion will :

* Take in a trained model , features (X) and labels (y)
* Create a meshgrid of the different X values 
* Make predictions across the meshgrid
* Plot the predictions as well as a line between zones (where each unique class falls) 


```python
import numpy as np

def plot_decision_boundary(model, X , y) :
    """
     Plots the decision boundary created by a model predicting on X .
    """
    
    # Define the axis boundaries of the plot and create a meshgrid
    x_min , x_max = X[: , 0].min() - 0.1 , X[: , 0].max() + 0.1
    y_min , y_max = X[: , 1].min() - 0.1 , X[: , 1].max() + 0.1
    
    xx , yy  = np.meshgrid(np.linspace(x_min , x_max , 100),
                           np.linspace(y_min , y_max , 100))
    
    # Create X values 
    x_in = np.c_[xx.ravel() , yy.ravel()] # Stack 2D arrays together
    
    # Make predictions
    y_pred = model.predict(x_in)
    
    # Check for milti-class
    if(len(y_pred[0]) > 1) :
        print("doing multiclass classification")
        # We have to reshape our prediction to get them ready for plotting
        y_pred = np.argmax(y_pred , axis=1).reshape(xx.shape)
    else :
        print("doing binary classification") 
        y_pred = np.round(y_pred).reshape(xx.shape)
        
    # Plot the decision boundary
    plt.contourf(xx , yy , y_pred , cmap=plt.cm.RdYlBu , alpha=0.7)
    plt.scatter(X[:,0] , X[:,1] , c=y , s=40 , cmap=plt.cm.RdYlBu)
    plt.xlim(xx.min() , xx.max())
    plt.ylim(yy.min() , yy.max())
        
```

```python
# Check the prediction our model is making
plot_decision_boundary(model=model_3 , X=X , y=y)
```

```python
plot_decision_boundary(model=model_2 , X=X , y=y)
```

```python
# Let's see if our model can be used for a regression problem ...

tf.random.set_seed(42)

# Create some regression data
X_regression = tf.range(0 , 1000 , 5)
y_regression = tf.range(100 , 1200 , 5)

X_regression , y_regression
```

```python
# Split our regression data into training and test sets 
X_reg_train = X_regression[:150]
X_reg_test = X_regression[150:]

y_reg_train = y_regression[:150]
y_reg_test = y_regression[150:]

# Setup random seed 
tf.random.set_seed(42)

# 1. Create the model
model_4 = tf.keras.Sequential([
    tf.keras.layers.Dense(100 , input_shape=(None,1)),
    tf.keras.layers.Dense(50),
    tf.keras.layers.Dense(1)
])

# 2. Compile the model
model_4.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.Adam(),
                metrics=["mae"])
 
# Fit our model to regression data
model_4.fit(X_reg_train , y_reg_train , epochs=100 )
```

```python
# Make predictions with our trained model
y_reg_pred = model_4.predict(X_reg_test)
```

```python
# Plot the model's predictions against the ground truth 
plt.figure(figsize=(10,7))
plt.scatter(X_reg_train[: , 0] , y_reg_train , c="b" , label="Training data")
plt.scatter(X_reg_test[: , 0] , y_reg_test , c="g" , label="Test data")
plt.scatter(X_reg_test[: , 0] , y_reg_pred , c="r" , label="Predictions")
plt.legend();
```
