### SweebApp - Dataset Preparation and Feature Engineering 📊🧠

In this phase of the project, three distinct datasets were selected and combined in order to build a more comprehensive and diverse data source. Each dataset contained network- and URL-related information, but with varying formats and feature distributions.

Two types of datasets will be created :

# 1. Structured Dataset (Tabular Data)

This version includes engineered features extracted from URLs (such as length, number of dots, and special characters) and will be used with classical machine learning models like Random Forest

To ensure consistency, all datasets were individually processed using custom preprocessing functions developed specifically for this project. These functions performed tasks such as:

 * `url_format()`: extracting structured features from raw URLs (e.g., length, number of dots, path length, protocol type) identifying suspicious patterns (e.g., presence of specific keywords, unusual characters)

 * `get_data_trainable()`: preprocess and transform a dataset into a structured format suitable for machine learning tasks, specifically for training and testing purposes

 * `create_final_dataset()`: standardizing the format across all datasets

Additionally, for datasets that included explicit attack-type labels, these were transformed into a unified classification scheme consisting of two categories: **Threat** OR **Safe**

# 2. Text-Based Dataset (NLP Dataset)

A second version of the dataset will retain the raw URL strings along with their corresponding labels. This dataset will be used for fine-tuning a transformer-based model (URLBERT) that operates directly on textual input and learns patterns automatically.


Finally, both datasets will be exported as .csv files and used in a subsequent phase of the project for model training and evaluation.

A comparative analysis will be conducted between **traditional machine learning models** & **transformer-based models**.

Datasets used :

```python
import operator as op
from urllib.parse import urlsplit
from sklearn.preprocessing import StandardScaler

top_tlds = ["com","cloud","shop","io","icu","info","pro","online","org","net","edu","co","ro"]
suspicious_words = ["login","verify","secure","account","update","bank","signin","password"]

def url_format(url):
  try:
    parsed_url = urlsplit(url)

    return {
      'url_length' : len(url),
      'num_dots' : url.count('.'),
      'num_slash' : url.count('/'),
      'https_protocol' : int(parsed_url.scheme == "https"),
      'num_special_char_domain' : str(parsed_url.netloc).count(r'[^A-Za-z0-9-]'),
      'suspicious_word_domain' : int(any(word in parsed_url.netloc.lower() for word in suspicious_words)),
      'top_tld' : int(parsed_url.netloc.split(".")[-1] in top_tlds),
      'len_path' : len(parsed_url.path),
      'suspicious_char_path' : int(any(word in parsed_url.path.lower() for word in suspicious_words))
    }
  except Exception:
    return None
```

```python
def get_data_trainable(data, url_column="url", columns=None,one_hot=None):
  '''
    This function will return data prepared for split into training and test.

    Params :
      - data : the Dataset to be transformed
      - columns : to remove some columns from the Dataset
      - one_hot : columns that need to be one-hot encoded
      - numeric_cols : columns that need to be normalized
      - url_column : the main column to apply 'url_format'
  '''
  if data is None:
    return

  data = data.dropna()
  if columns is not None:
    data = data.drop(columns=columns)

  if data[url_column] is not None:
    urls_data = data[url_column].apply(url_format).apply(pd.Series)

  if one_hot is not None:
    data = pd.get_dummies(data, columns=one_hot, drop_first=True)

  data = data.drop(url_column, axis='columns')

  df_final = pd.concat([data,urls_data], axis=1)

  return df_final
```

```python
threat_types = ["threat","safe"]
main_cols = ['suspicious_char_path','len_path','top_tld','suspicious_word_domain','num_special_char_domain','https_protocol','num_slash','num_dots','url_length']

def create_final_dataset(dataset, detection_cols=None):
  '''
    This function standardize all datasets to a common data type.

    Params :
      - dataset : Set the dataset to be transformed
      - detection_cols : Insert here columns wich scope is to detect if is threat or not
  '''

  if dataset is None:
    return


  if not detection_cols:
    print("Please provide detection_cols")

  main_data = dataset[main_cols].copy()
  feature_change = dataset[detection_cols].copy()
  feature_change["is_threat"] = feature_change[detection_cols].any(axis=1).map({True : "threat", False : "safe"})
  feature_change = feature_change["is_threat"]
  final_dataset = pd.concat([main_data, feature_change],axis=1)

  return final_dataset

```

```python
threat_types = ["threat","safe"]

def create_final_nlp_dataset(dataset, detection_cols=None, main_cols=None):
  '''
    This function standardize all datasets to a common data type.

    Params :
      - dataset : Set the dataset to be transformed
      - detection_cols : Insert here columns wich scope is to detect if is threat or not
      - main_cols : Insert here columns wich scope is to keep
  '''

  if dataset is None:
    return


  if not detection_cols:
    print("Please provide detection_cols")

  if not main_cols:
    print("Please provide main_cols")

  main_data = dataset[main_cols].copy()
  feature_change = dataset[detection_cols].copy()
  feature_change["is_threat"] = feature_change[detection_cols].any(axis=1).map({True : 1, False : 0})
  feature_change = feature_change["is_threat"]
  final_dataset = pd.concat([main_data, feature_change],axis=1)

  return final_dataset
```
