## Creating random tensors

Random tensors are tensors of some abitrary size which contain random numbers .

**Matrix multiplication**

In machine learning , matrix multiplication is one of the most common tensor operations .

There are two rules our tensors (or matrices) need to fulfil if we're going to matrix multiply them :

1. The inner dimensions must match
2. The resulting matrix has the shape of the inner dimensions

```python
# Matrix multiplication in tensorflow

print(tensor)
tf.matmul(tensor , tensor)
```

```python
# Matrix multiplication with Python operator "@"
tensor @ tensor
```

```python
tensor.shape
```

```python
# Create a tensor of (3,2)
X = tf.constant([[1,2] , [3,4] , [5,6]])
# Create another (3,2) tensor
Y = tf.constant([[7,8] , [9,10] , [11,12]])

X , Y
```

```python
# Let's change the shape of Y
Y_reshaped = tf.reshape(Y , shape=(2,3))
X @ Y_reshaped
```

```python
X.shape , Y_reshaped.shape
```

```python
# Try to multiply X by reshaped Y
X @ tf.reshape(Y , shape=(2,3))
```

```python
tf.matmul(X , tf.reshape(Y , shape=(2,3)))
```

```python
tf.matmul(tf.reshape(X , shape=(2,3)) , Y)
```

```python
# Can do the same with transpose
X , tf.transpose(X) , tf.reshape(X , shape=(2,3))
```

```python
# Try matrix multiplication with transpose  rather than reshape
tf.matmul(tf.transpose(X) , Y)
```

**The dot product**

Matrix multiplication is also reffered to as the dot product .

You can perform matrix mutiplication using :
* `tf.matmul()`
* `tf.tesnordot()`

```python
X , Y
```

```python
# Perfom the dot porduct o X and Y (require X or Y to be transported)
tf.tensordot(tf.transpose(X) , Y , axes=1)
```

```python
# Perfom matrix multiplication between X and Y (transposed)
tf.matmul(X , tf.transpose(Y))
```

```python
# Perfom matrix multiplication between X and Y (reshaped)
tf.matmul(X , tf.reshape(Y , shape=(2,3)))
```

```python
# Check the values of Y , reshape Y and transposed Y
print("Normal Y : ")
print(Y , "\n") # '\n' is for newline

print("Y reshaped to (2,3) : ")
print(tf.reshape(Y , (2,3)) , "\n")

print("Y transposed : ")
print(tf.transpose(Y))
```

```python
tf.matmul(X , tf.transpose(Y))
```

### Aggregating tensors

Aggregating tensors = condensing them from multiple values down to a smaller amount of values

```python
# Get the absolute values
D = tf.constant([-7 , -10])
D
```

```python
# Get absolute values
tf.abs(D)
```

Following form of Aggregation :

* Get the minimum
* Get the maximum
* Get the mean of a tensor
* Get the sum of a tensor




```python
# Creating a random tensor with values between 0 and 100 of size 50
E = tf.constant(np.random.randint(0,100,size=50))
E
```

```python
tf.size(E) , E.shape , E.ndim
```

```python
# Find the minimum
tf.reduce_min(E)
```

```python
# Find the maximum
tf.reduce_max(E)
```

```python
# Find the mean
tf.reduce_mean(E)
```

```python
# Find the sum
tf.reduce_sum(E)
```

```python
E_float = tf.cast(E , dtype=tf.float32)
```

```python
# Variants - find variance of our tensor
E_variance = tf.math.reduce_variance(E_float)
E_variance
```

```python
import tensorflow_probability as tfp
tfp.stats.variance(E_float)
```

```python
# Deviation - find deviation of our tensor
tf.math.reduce_std(E_float)
```

## Find the positional maximum and minimum of a tensor

```python
# Create a new tensor for finding positional minimum and maximum
tf.random.set_seed(42)
F = tf.random.uniform(shape=[50])
F
```

```python
# Find the positional maximum
tf.argmax(F)
```

```python
# Index on our largest value position
F[tf.argmax(F)]
```

```python
# Find the max value of F
tf.reduce_max(F)
```

```python
# Check for equality
F[tf.argmax(F)] == tf.reduce_max(F)
```

```python
 # Find the positional minimum
tf.argmin(F)
```

```python
# Index on our minimum value position
F[tf.argmin(F)]
```

```python
# Find the min value of F
tf.reduce_min(F)
```

```python
# Check for equality
F[tf.argmin(F)] == tf.reduce_min(F)
```
