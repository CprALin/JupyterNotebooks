### Dot product example (nut butter sales)

```python
np.random.seed(0)
sales_amounts = np.random.randint(20 , size=(5,3))
sales_amounts
```

```python
# Create weekley_sales DataFrame
weekly_sales = pd.DataFrame(sales_amounts , index=["Mon" , "Tues" , "Wed" , "Thurs" , "Fri"] , columns=["Almound butter" , "Peanut butter", "Cashew butter"])
weekly_sales
```

```python
# Create prices array
prices = np.array([10,8,12])
prices
```

```python
# Create butter_prices DataFrame
butter_prices = pd.DataFrame(prices.reshape(1,3) , index=["Price"] , columns=["Almound butter" , "Peanut butter", "Cashew butter"])
butter_prices
```

```python
sales_amounts.shape , prices.shape
```

```python
# Shape aren't aligned , let's transpose
total_sales = prices.dot(sales_amounts.T)
total_sales
```

```python
# Create daily_sales
butter_prices.shape , weekly_sales.shape
```

```python
daily_sales = butter_prices.dot(weekly_sales.T)
daily_sales
```

```python
weekly_sales["Total ($)"] = daily_sales.T
weekly_sales
```
